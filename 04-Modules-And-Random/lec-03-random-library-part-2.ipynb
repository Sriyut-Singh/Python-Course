```python
print('you have two dice options\n1.1 will show only result of dice\n2.2 will show two results of dice ')
while True:
    c=int(input('write number as your choice of dice: '))
    if c==1:
        print(random.randint(1,6))
    elif c==2:
        print(random.randint(1,6))
        print(random.randint(1,6))
    else:
        print('game over')
        break
```

```python
print('welcom to game')
print('your guess should be in between 1 to 10')
b=random.randint(1,10)
print('you have only three chances')
for i in range(3):
    print('you have' ,(3-i), 'chances remaning')
    a=int(input('write your number: '))
    print('your guess is',a)
    if a>b:
        print('your guess is bigger than the number')
    if a<b:
        print('your guess is less than the number')
    elif a==b:
        print('your guess is correct')
        break
print('thanks for playing the game')
```
