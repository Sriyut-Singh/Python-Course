# Starting the Debugger

### From the Command Line

```python
def seq(n):
    for i in range(n):
        print(i)
    return

seq(6)
```

```python
import pdb 

#interactive debugging
def seq(n):
    for i in range(n):
        pdb.set_trace() #breakpoint
        print(i)
    return

a = seq(5)
```
