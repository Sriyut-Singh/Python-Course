```python
import numpy as np
import pandas as pd
#Create DataFrame
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
       'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
       'Sales':[200,120,340,124,243,350]}
df = pd.DataFrame(data)
bycomp = df.groupby('Company')
print('\nGrouping the DataFrame by "company" column and listing mean sales\n',
     '-'*55,
     sep='')
print(bycomp.mean())#we can find any thing of this
```

```python
#testing
print('\nGrouping the DataFrame by "company" column and listing median sales\n',
     '-'*55,
     sep='')
print(bycomp.median())
print('\nGrouping the DataFrame by "company" column and listing sum sales\n',
     '-'*55,
     sep='')
print(bycomp.sum())
```

```python
#merging by a common key
left = pd.DataFrame({'key': ['K0', 'K8', 'K2', 'K3'],
                     'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']})
   
right = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                          'C': ['C0', 'C1', 'C2', 'C3'],
                          'D': ['D0', 'D1', 'D2', 'D3']})
```

```python
left
```

```python
right
```

```python
merge1 = pd.merge(left,right,
                  how='inner',#inner means first written in our case it is left
                 on='key')
merge1
```

```python
merge2 = pd.merge(left,right,
                  how='outer',#outer means first written in our case it is right
                 on='key')
merge2
```

```python
left = pd.DataFrame({'key1': ['K0', 'K0', 'K1', 'K2'],
                     'key2': ['K0', 'K1', 'K0', 'K1'],
                        'A': ['A0', 'A1', 'A2', 'A3'],
                        'B': ['B0', 'B1', 'B2', 'B3']})
    
right = pd.DataFrame({'key1': ['K0', 'K1', 'K1', 'K2'],
                               'key2': ['K0', 'K0', 'K0', 'K0'],
                                  'C': ['C0', 'C1', 'C2', 'C3'],
                                  'D': ['D0', 'D1', 'D2', 'D3']})
```

```python
merge3 = pd.merge(left,right,
                  how='left',
                 on=['key1','key2'])
merge3
```

```python
merge3 = pd.merge(left,right,
                  how='right',
                 on=['key1','key2'])
merge3
```

```python
merge4 = pd.merge(left,right,#will show only common as default is inner
                 on=['key1','key2'])
merge4
```
