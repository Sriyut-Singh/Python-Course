```python
import random
list1=["jazz","hello","friends","mother","father"]
word=random.choice(list1)
guesses=" "
turns=6
while turns>0:
    failed=0
    for x in word:
        if x in guesses:
            print(x,end=" ")
        else:
            print("_",end=" ")
            failed+=1
    print(" ")
    if failed==0:
        print("You Win")
        print("The word is: " , word)
        break
    guess=input("Enter your guess: ")
    if guess not in word or guess in guesses:
        turns-=1
        print("Wrong")
        print("You have",turns,"guesses left.")
        if turns==0:
            print("You lose")
    guesses+=guess
```
