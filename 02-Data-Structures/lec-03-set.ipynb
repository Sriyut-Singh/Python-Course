# HOME WORK

```python
hw={1,2,34,5,6,7,8,9,'a',24,39,5,9,6,7}#making a set

work={3,5,7,8,2,1,4,6764,34,474}#another set 

hw.isdisjoint('s')#not takes integer object

print('your set is=',hw)
print('your another set is=',work)

print('doing issubset=',hw.issubset(work))
#check whether another set contains this set.

print('doing symmetric_difference=',hw.symmetric_difference(work))
#Return the symmetric difference of two sets as a new set

print('doing difference=',hw.difference(work))
#Return the difference of two or more sets as a new set.

print('doing issuperset=',hw.issuperset(work))
#Report whether this set contains another set.

print('doing difference_update=',hw.difference_update(work))
#Remove all elements of another set from this set.

print('doing symmetric_difference_update',hw.symmetric_difference_update(hw))
#Update a set with the symmetric difference of itself and another.
print(hw)
#nothing is left in the set.
```
