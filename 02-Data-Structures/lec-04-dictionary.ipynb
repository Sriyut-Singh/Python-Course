```python
# iterating

firefox={'name':'SRIYUT KUMAR','age':20,'salary':80000,'company':'FIREFOX','year':2018}
for i in firefox:
    print(i)
```

```python
firefox={'name':'SRIYUT KUMAR','age':20,'salary':80000,'company':'FIREFOX','year':2018}
for i in firefox.values():
    print(i)
```

```python
firefox={'name':'SRIYUT KUMAR','age':20,'salary':80000,'company':'FIREFOX','year':2018}
for x,i in firefox.items():
    print(x,i)
```
