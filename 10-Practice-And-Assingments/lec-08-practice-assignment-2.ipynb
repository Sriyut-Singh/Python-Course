# Create the below pattern using nested for loop in Python.

# *
# **
# ***
# ****
# *****
# ****
# ***
# **
# *

```python
for i in range(1):
    for i in range(5):
        print('*'*i)
    for i in range(5):
        print('*'*(5-i))
```

Write a Python program to reverse a word after accepting the input from the user.

Input word: ineuron

Output: norueni

```python
a = input('Word: ')
print('Reverse=',a[::-1])
```
