# Trying to make hang man game

```python
import random#random library is imported 

list1=["jazz","hello","friends","mother","father"]#a list for the guessing word


word=random.choice(list1)#computer will choose any letter


guesses=" "#for the enpty string if answer is not correct
turns=6#for the chances which will decrease
while turns>0:#while loop for the guess
    failed=0#string to increase when we are failed
    for x in word:#for loop for the computer choice
        if x in guesses:#if it is correct then print the letter at the position 
            print(x,end=" ")
        else:#if the answer is wrong so end statement
            print("_",end=" ")
            failed+=1#if answer is not correct the failed value will increase
    print(" ")#for  a new line
    if failed==0:#because the failed vale will be zero you will win 
        print("You Win")
        print("The word is: " , word)
        break#then it will break the game
    guess=input("Enter your guess: ")#we have to wrie our guess
    if guess not in word or guess in guesses:#for checking while it is in the string or not 
        turns-=1#if not then  it  will do  -1
        print("Wrong")
        print("You have",turns,"guesses left.")
        if turns==0:#when turns get fully decreased you lose will be printed 
                    #means when guess will be zero(0)
            print("You lose")
    guesses+=guess
```
