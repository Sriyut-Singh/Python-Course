```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
a = sns.load_dataset('flights')
```

```python
a.head()
```

```python
sns.relplot(x='passengers',y='month',data=a)#we can make a graph by three column 
```

```python
sns.relplot(x='passengers',y='month',hue='year',data=a)#hue changes the colour
```

```python
b = sns.load_dataset('tips')
```

```python
b.head()
```

```python
sns.relplot(x='time',y='tip',data=b,kind='line')
```

```python
sns.catplot(x='day',y='total_bill',data=b)
```

```python
sns.catplot(x='day',y='total_bill',data=b,kind='violin')
```

```python
c=sns.load_dataset('iris')
d=sns.FacetGrid(c,col='species')
d.map(plt.hist,'sepal_length')
```
