# Line plot

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
x = np.arange(1,11)
x
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
plt.plot(x,y)
plt.show()#it is important to write show to show it if it is in between
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
plt.plot(x,y)
plt.title('line plot')
plt.xlabel("X-lable")
plt.ylabel("y-lable")
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
plt.plot(x,y,color='red',linestyle='-.',linewidth=2)
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
x=np.arange(1,11)
y1=2*x
y2=3*x
plt.plot(x,y1)
plt.plot(x,y2)
plt.show()
```

#### There is no lable. We can add it.

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
plt.plot(x,y1,color='g',linestyle='--',linewidth=2,label='line1')
plt.plot(x,y2,color='r',linestyle='-.',linewidth=3,label='line2')
plt.title("two Line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend() #include legend line 1 and line 2
plt.grid(True)#it will show the grid before
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
x=np.arange(1,11)
y1=2*x
y2=3*x
plt.subplot(2,1,1)
plt.plot(x,y1)
plt.subplot(2,1,2)
plt.plot(x,y2)
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
x=np.arange(1,11)
y1=2*x
y2=3*x
plt.subplot(1,2,1)
plt.plot(x,y1)
plt.subplot(1,2,2)
plt.plot(x,y2)
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
x=np.arange(1,11)
y1=2*x
y2=3*x
plt.subplot(2,1,1)
plt.plot(x,y1,color='g',linestyle=':',linewidth=2)
plt.title("two Line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.subplot(2,1,2)
plt.plot(x,y2,color='r',linestyle='--',linewidth=1)
plt.title("two Line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```

# Bar plot

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
student = {'BoB':87,'Julia':76,'Anne':99,'Matt':45}
name = list(student.keys())
name
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
student = {'BoB':87,'Julia':76,'Anne':99,'Matt':45}
marks = list(student.values())
marks
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
plt.bar(name,marks)
plt.title('Bar plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
plt.barh(name,marks,color='y')
plt.title('Bar h plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

# Pie chart

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
fruits = ['APPLE','MANGO','GUAVA','ORANGE','PINEAPPLE']
quantity = [10,30,11,9,19]
plt.pie(quantity,labels=fruits)
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
fruits = ['APPLE','MANGO','GUAVA','ORANGE','PINEAPPLE']
quantity = [10,30,11,9,19]
plt.pie(quantity,labels=fruits,autopct='%0.1f%%',colors=['yellow','blue','red','orange','green'])
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
fruits = ['APPLE','MANGO','GUAVA','ORANGE','PINEAPPLE']
quantity = [10,30,11,9,19]
plt.pie(quantity,labels=fruits,autopct='%0.1f%%',colors=['yellow','blue','red','orange','green'],explode=(0,0.1,0,0,0))# to bring it out
plt.show()
```

# DoughNut chart

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
fruits = ['APPLE','MANGO','GUAVA','ORANGE','PINEAPPLE']
quantity = [10,30,11,9,19]
plt.pie(quantity,labels=fruits,radius=2)
plt.pie([1],colors=['w'],radius=1)#color of inner circle white
plt.show()
```

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
fruits = ['APPLE','MANGO','GUAVA','ORANGE','PINEAPPLE']
quantity = [10,30,11,9,19]
plt.pie(quantity,labels=fruits,autopct='%0.1f%%',radius=2)
plt.pie([1],colors=['w'],radius=1)#color of inner circle white
plt.show()
```
