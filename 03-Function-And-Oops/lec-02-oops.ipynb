```python
class employee:
    #class attribute
    company="Google"
    def __init__(self,name,age,salary):
        #instance attribute
        self.name = name
        self.age = age
        self.salary = salary
        
# creting object
emp= employee('Sriyut',15,100000)
emp1=employee('king',16,100000)

#accessing class attributes using __class__ method
# syntax is-- instance.__class__.attribute
print(f"{emp.name} and {emp1.name} work for {emp.__class__.company}")

#accesing instance attributes
print(f"{emp.name}'s age is {emp.age} and salary is {emp.salary}")
print(f"{emp1.name}'s age is {emp1.age} and salary is {emp1.salary}")
```

```python
class Parrot:
    
    # instance attributes
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    # instance method
    def sing(self, song):
        return "{} sings {}".format(self.name, song)

    def dance(self):
        return "{} is now dancing".format(self.name)

# instantiate the object
blu = Parrot("Blu", 10)

# call our instance methods
print(blu.sing("'Happy'"))
print(blu.dance())
```

```python
class bird:
    
    def __init__(self):
        print('Bird is ready')
        
    def whoisthis(self):
        print('Bird')
    
    def swim(self):
        print('Swim faster')
        
#child class
class penguin(bird):
    
    def __init__(self):
        #call super() function for other class init
        super().__init__()
        print('Penguin is ready')
        
    def whoisthis(self):
        print('Penguin')
    
    def run(self):
        print('Run faster')
        
peggy = penguin()
peggy.whoisthis()
peggy.swim()
peggy.run()
```

### Multi-level Inheritence

```python
class grandfather:
    def __init__(self,grandfather_name):
        self.grandfather_name = grandfather_name

class dad(grandfather):
    def __init__(self,father_name,grandfather_name):
        self.father_name = father_name
        
        grandfather.__init__(self,grandfather_name)
        
class son(dad):
    def __init__(self,son_name,father_name,grandfather_name):
        self.son_name = son_name
        
        dad.__init__(self,father_name,grandfather_name)
        
    def print_name(self):
        print('Grandfather name =',self.grandfather_name)
        print('Father name = ',self.father_name)
        print('Son name =',self.son_name)
        
s1 = son('Lucky', 'Raj', 'Kamal')
s1.print_name()
```

# Encapsulation

```python
class computer:
    
    def __init__(self):
        #to Encapsulate use '_' or '__'
        self.__maxprice = 900
        
    def sell(self):
        print('Selling price: {}'.format(self.__maxprice))
    
    def SetMaxPrice(self, price):
        self.__maxprice = price
        
c = computer()
c.sell()

#changing the price
c.__maxprice = 100
c.sell()

#using setter function
c.SetMaxPrice(1000)
c.sell()
```

# Polymorphism

```python
class Parrot:

    def fly(self):
        print("Parrot can fly")
    
    def swim(self):
        print("Parrot can talk")

class Penguin:

    def fly(self):
        print("Penguin can swim")
    
    def swim(self):
        print("Penguin can swim")

# common interface
def flying_test(bird):
    bird.fly()
    bird.swim()
#instantiate objects
blu = Parrot()
peggy = Penguin()

# passing the object
flying_test(blu)
flying_test(peggy)
```
